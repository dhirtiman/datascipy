```python
# 2 
# 2. Implement the Five Steps of Data Science with a small dataset 
# (import - clean - analyze -
# visualize - interpret).
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df=sns.load_dataset("iris")

print("\n first 5 rows pf the datasets")
print(df.head())

print("\n Dataset shape:")
print(df.shape)

print("\n Column name:")
print(df.columns.tolist())


print("Missing values BEFORE cleaning")

print(df.isnull().sum())

print("\n Number of suplicate records BEFORE cleaning:")
print(df.duplicated().sum())


print("Identify categorical columns")

categorical_columns= df.select_dtypes(include="object").columns

numeric_columns=df.select_dtypes(include="number").columns
print("fill numerical null vaules with the columns mean")

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())

df=df.drop_duplicates()

print("\nMissing values AFTER cleaning")
print(df.isnull().sum())

print("\nNumber of duplicate records AFTER cleaning")

print(df.duplicated().sum())

print("\n Dataset Shape AFTER cleaning")
print(df.shape)


```

```python
print("\n=====STEP 3 ANALYZE=====")

print("\n statistical summary:")

print(df.describe())

print("\nAverage measurement for each species: ")
print(df.groupby("species").mean(numeric_only=True))




```

```python
print("\nNumber of flowers in each species: ")
print(df["species"].value_counts())
```

```python
print("\nMaximum petal length: ")
print(df["petal_length"].max())
```

```python
print("\nMinimum petal length: ")
print(df["petal_length"].min())
```

```python
print("\n=== STEP 4: VISUALIZE === ")

plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="sepal_length",
    y="petal_length",
    hue="species"
)
plt.title("sepal length vs petal length")

plt.xlabel("sepal length(cm)")
plt.ylabel("petal length(cm)")

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="sepal_width",
    y="petal_width",
    hue="species"
)

plt.title("sepal width vs petal width")

plt.xlabel("sepal width(cm)")
plt.ylabel("petal width(cm)")

plt.show()
```
