```python
import pandas as pd
from sklearn.datasets import load_iris
from typing import cast
from sklearn.utils import Bunch
from sklearn.preprocessing import (
    MinMaxScaler,StandardScaler,
    RobustScaler,MaxAbsScaler,Normalizer
)
# 

# load dataset
iris = load_iris()
iris = cast(Bunch, load_iris())
iris.data # pyright: ignore[reportAttributeAccessIssue]
df = pd.DataFrame(iris.data,columns=iris.feature_names)
print("Original DataSet")
print(df.head())
```

```python
#1.Mix-Max Scaling
minmax = MinMaxScaler()
minmax_data=minmax.fit_transform(df)
#2.Standardization
standard = StandardScaler()
standard_data=standard.fit_transform(df)
#3.Robust Scaling
robust = RobustScaler()
robust_data=robust.fit_transform(df)
#4.Max Absolute Scaling
maxabs = MaxAbsScaler()
maxabs_data=maxabs.fit_transform(df)

#L1 Normalization

l1 = Normalizer(norm='l1')
l1_data=l1.fit_transform(df)

#L2 Normalization
l2 = Normalizer(norm='l2')
l2_data=l2.fit_transform(df)

# print(pd.DataFrame(l1_data,columns=df.columns).head())
# print(pd.DataFrame(l2_data,columns=df.columns).head())


results = {
    "Mix-Max Scaling": minmax_data,
    "Standardization": standard_data,
    "Robust Scaling": robust_data,
    "Max Absolute Scaling": maxabs_data,
    "L1 Normalization": l1_data,
    "L2 Normalization": l2_data,
}

for name,values in results.items():
    print("\n",name)
    print(pd.DataFrame(values,columns=df.columns).head())
```
