```python
#4. Import a dataset using Pandas and display descriptive statistics 
# (mean, median, mode, etc.)
import pandas as pd

# Step 1: Import the Iris dataset
df = pd.read_csv("iris.csv")

# Step 2: Display the first five records
print("FIRST FIVE RECORDS")
print(df.head())

# Step 3: Select numerical attributes
numeric_df = df.select_dtypes(include="number")

# Step 4: Display descriptive statistics
print("\nDESCRIPTIVE STATISTICS")
print(numeric_df.describe())

# Step 5: Calculate and interpret statistics
print("\nINTERPRETATION OF DESCRIPTIVE STATISTICS")

for column in numeric_df.columns:
    data = numeric_df[column]

    mean = data.mean()
    median = data.median()
    mode = data.mode().iloc[0]
    minimum = data.min()
    maximum = data.max()
    data_range = maximum - minimum
    variance = data.var(ddof=0)
    std_dev = data.std(ddof=0)

    print(f"\nAttribute: {column}")
    print(f"Mean: {mean:.2f}")
    print(f"Median: {median:.2f}")
    print(f"Mode: {mode:.2f}")
    print(f"Minimum: {minimum:.2f}")
    print(f"Maximum: {maximum:.2f}")
    print(f"Range: {data_range:.2f}")
    print(f"Population Variance: {variance:.2f}")
    print(f"Population Standard Deviation: {std_dev:.2f}")

    print("\nInterpretation:")

    print(
        f"- Mean: The average {column} is {mean:.2f}."
    )

    print(
        f"- Median: Half of the observations are at or below "
        f"{median:.2f}, and half are at or above it."
    )

    print(
        f"- Mode: The most frequently occurring value "
        f"is {mode:.2f}."
    )

    print(
        f"- Range: The values span {data_range:.2f} units, "
        f"from {minimum:.2f} to {maximum:.2f}."
    )

    print(
        f"- Variance: The population variance is {variance:.2f}, "
        f"indicating the squared spread around the mean."
    )

    print(
        f"- Standard Deviation: The typical spread around the "
        f"mean is summarized by {std_dev:.2f} units."
    )

    if std_dev < 0.5:
        print("- Spread: The values show relatively low variability.")
    elif std_dev < 1.0:
        print("- Spread: The values show moderate variability.")
    else:
        print("- Spread: The values show relatively high variability.")
```
