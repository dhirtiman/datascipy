## list

```python
shop = ['apple','banana','orange']
print(shop)
# append
shop.append('grape')
shop.extend(['mango','grapefruit'])
print(shop)
```

```python
shop2 = ['a','b','c','d']
print(shop2)

flist = shop+shop2
print(flist)
```

```python
# indexing 
print(flist[0])
print(flist[-1])
print(flist[2][3])
```

```python
#  slicing
flist = shop+shop2
print(flist)
print(flist[0:3:1])
print(flist[:3])
print(flist[3:])
print(flist[::])
print(flist[::2])
print(flist[::-1])
```

```python
# sorting
flist.sort()
print(flist)
```

```python
flist[0]='sivadev'
flist[2]='swetha'
flist[4]='sruthi'
flist.remove('d')
flist.pop(-1)
print(flist)
```

```python
nlist = flist.copy()
print(len(nlist))
print(nlist)
del flist[5::]
print(flist)
```

```python
flist.insert(-1,'orange')
print(flist)
```

```python
# reverse
flist.reverse
print(flist)
```

```python
print(flist*10)
print(flist.count('banana'))
```
