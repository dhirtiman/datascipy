## SETS

```python
s = {1,2,3,4,5}
print(s)
print(type(s))
```

```python
s.add(3)
print(s)
s.add(8)
print(s)
```

```python
# copy
s2 = s.copy()
print(s2)
s.clear()
print(s)
```

```python
shop = {'milk','bread','cheese','butter'}
shop2 = {'pencil','pen','eraser','butter'}
print(shop.difference(shop2))
print(shop2.difference(shop))
```

```python
s = {1,2,3,4,5}
# discard() remove the specific item from the set
s.discard(2)
print(s)
s.remove(5)
print(s.__len__())
```

```python
print(s.pop())
```

```python
s = {1,2,3,4}
s1 = {3,4,5,6}
s.difference_update(s1)
print(s)
```

```python
s2 = {1,2,3,4}
s1 = {3,4,5,6}
s3 = s1.intersection(s2)
print(s3)
```

```python
s2 = {1,2,3,4}
s1 = {3,4,5,6}
s3 = s1.union(s2)
print(s3)
```

```python
s2 = {1,2,3,4}
s1 = {3,4,5,6}
s1.update(s2)
print(s1)
```
