# Dictionary 
##### here is some code about dictionary
* import the dictionary module
* create a dictionary
* add items to the dictionary
* print the dictionary
* print the keys


```python
my_dict = {
    'name': 'Bigie',
    'age': 55,
    'gender': 'male'
}

print(my_dict)
my_dict.clear()
print('after clearing',my_dict)

my_dict = {
    'name': 'Bigie',
    'age': 55,
    'gender': 'male'
}
print(my_dict)
new_dic = my_dict.copy()
print(new_dic)
print(my_dict==new_dic)
print(my_dict is new_dic)

# get()

```

```python
# popitem()
print(new_dic.popitem())
print(new_dic)
# update
new_dic.update({'age': 60,'gender': 'male'})
print(new_dic)
# del
del new_dic['name']
print(new_dic)
```
