```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import sklearn
import math
```

```python
# create pizza dataset

data = {
    "Diameter":[6,8,10,12,14],
    "Price":[100,130,160,190,220]
    }

df=pd.DataFrame(data)
```

```python
print(df)
```

```python
plt.scatter(df["Diameter"],df["Price"])
plt.xlabel("Diameter")
plt.ylabel("Price")
plt.show()
```

```python
# seperate the data into independent and dependent
x=df[["Diameter"]] #scikit learn expect x to be two dimensional
y=df["Price"]

print(x)
print(y)
```

```python
# training the model
# y=mx+c

model = LinearRegression()
model.fit(x,y)

```

```python
val = model.predict([[16]])
print(val)
```

```python
print("slope m:",math.floor(model.coef_[0]))
```

```python
print("intercept c:",math.ceil(model.intercept_))
```

```python
new_data = pd.DataFrame({
    "Diameter":[10,16,17,21,25],
})


predict_price = model.predict(new_data)
print(predict_price)
```

```python
plt.scatter(x,y,label="Actual data")
plt.plot(x["Diameter"],model.predict(x),color="green",label="regression LIne")
plt.grid(True)
plt.legend()
plt.show()
```
