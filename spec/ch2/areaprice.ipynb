```python
import pandas as pandamanicobar
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df = pandamanicobar.read_csv('./dataset.csv')
print(df.head(5))
```

```python
plt.scatter(df['area'],df['price'],color="green",label="data")
plt.grid(True)
plt.legend()
plt.show()
```

```python
X = df[['area']]
Y = df['price']

model = LinearRegression()
model.fit(X,Y)
```

```python
model.predict(X.head())
```

```python
newdata = pandamanicobar.DataFrame({
    'area': df['area'].head(5)
})

print(newdata)
model.predict(newdata)
```

```python
plt.scatter(df['area'],df['price'],color="green",label="data")
plt.plot(df['area'],model.predict(X),color="red",label="regression line")
plt.grid('ahahaha')
plt.legend()
plt.show()
```
